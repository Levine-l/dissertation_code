```python
#Extract the dataset archive.
import os
zip_path = '/content/drive/MyDrive/Dissertation/ukceh_east_anglia.zip'
extract_path = '/content/drive/MyDrive/Dissertation/'
if os.path.exists(zip_path):
    !unzip -o "{zip_path}" -d "{extract_path}"
    print(f"Extracted to: {extract_path}")
else:
    print(f"File not found: {zip_path}")
```

```python
#check extracted files
target_dir = '/content/drive/MyDrive/Dissertation/ukceh_east_anglia'
readme_path = os.path.join(target_dir, 'README_inventory.md')
print("Directory contents:")
for root, dirs, files in os.walk(target_dir):
    level = root.replace(target_dir, '').count(os.sep)
    indent = ' ' * 4 * (level)
    print(f"{indent}{os.path.basename(root)}/")
    sub_indent = ' ' * 4 * (level + 1)
    for f in files:
        print(f"{sub_indent}{f}")
print("\n--- README_inventory.md ---")
if os.path.exists(readme_path):
    with open(readme_path, 'r', encoding='utf-8') as f:
        print(f.read())
else:
    print("README_inventory.md not found.")
```

```python
#check the UKCEH GeoPackage schema
import geopandas as gpd
gpkg_file = '/content/drive/MyDrive/Dissertation/ukceh_east_anglia/lccm-2022_6395063/lccm-2022_6395063.gpkg'
if os.path.exists(gpkg_file):
    try:
        gdf = gpd.read_file(gpkg_file, rows=5)
        print("Loaded data:")
        display(gdf)
        print("\nColumns:")
        print(gdf.columns.tolist())
        if 'crop_code' in gdf.columns:
            print("\n'crop_code' found.")
        else:
            print("\n'crop_code' not found.")
    except Exception as e:
        print(f"Read failed: {e}")
else:
    print("Invalid file path.")
```
